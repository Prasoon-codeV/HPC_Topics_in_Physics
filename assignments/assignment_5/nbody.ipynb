```python
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
```

# Serial Method

```python
N = [128, 256, 512, 1024, 2048, 4096]
serial_times = [.144, .222, .542, 1.572, 6.259, 23.545]

plt.figure(figsize=(10, 6))
plt.plot(N, serial_times, marker='o', label='Serial Time')
plt.xscale('log', base=2)
plt.yscale('log')
plt.xlabel('Number of Masses (N)')
plt.ylabel('Execution Time (seconds)')
plt.title('Execution Time vs Number of Masses')
plt.grid(True, which="both", ls="--")
plt.legend()
```

# OpenMP Parallelization

```python
N = 4096
ranks = [1, 2, 4, 8, 16, 32]
omp_times = [18.452, 9.263, 4.654, 2.362, 1.247, 1.241]

omp_run_time = [i/N**2 for i in omp_times]
plt.figure(figsize=(10, 6))
plt.plot(ranks, omp_run_time, marker='o', label='OMP Time')
plt.xscale('log', base=2)
plt.yscale('log')
plt.xlabel('Number of Threads')
plt.ylabel('Runtime/N^2 (seconds)')
plt.title('Runtime/N^2 vs Number of Threads for N=4096')
plt.grid(True, which="both", ls="--")
plt.legend()
```

# MPI Parallelization

```python
N = 4096
ranks = [1, 2, 4, 8, 16, 32]
mpi_times = [20.469, 11.317, 6.802, 4.326, 3.17, 4.241]
mpi_run_time = [i/N**2 for i in mpi_times]

plt.figure(figsize=(10, 6))
plt.plot(ranks, mpi_run_time, marker='o', label='MPI Time')
plt.xscale('log', base=2)
plt.yscale('log')
plt.xlabel('Number of Ranks')
plt.ylabel('Runtime/N^2 (seconds)')
plt.title('Runtime/N^2 vs Number of Ranks for N=4096')
plt.grid(True, which="both", ls="--")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(ranks, omp_run_time, marker='o', label='OMP Time')
plt.plot(ranks, mpi_run_time, marker='o', label='MPI Time')
plt.xscale('log', base=2)
plt.yscale('log')
plt.xlabel('Number of Threads/Ranks')
plt.ylabel('Runtime/N^2 (seconds)')
plt.title('Runtime/N^2 vs Number of Threads/Ranks for N=4096')
plt.grid(True, which="both", ls="--")
plt.legend()
plt.show()
```

# Shared MPI

```python
N = 4096
shared_mpi_times = [31.664, 17.302, 10.085, 6.517, 4.127, 4.304]
shared_mpi_run_time = [i/N**2 for i in shared_mpi_times]

plt.figure(figsize=(10, 6))
plt.plot(ranks, shared_mpi_run_time, marker='o', label='Shared MPI Time')
plt.xscale('log', base=2)
plt.yscale('log')
plt.xlabel('Number of Ranks')
plt.ylabel('Runtime/N^2 (seconds)')
plt.title('Runtime/N^2 vs Number of Ranks for Shared MPI (N=4096)')
plt.grid(True, which="both", ls="--")
plt.legend()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(ranks, omp_run_time, marker='o', label='OMP Time')
plt.plot(ranks, mpi_run_time, marker='o', label='MPI Time')
plt.plot(ranks, shared_mpi_run_time, marker='o', label='Shared MPI Time')
plt.xscale('log', base=2)
plt.yscale('log')
plt.xlabel('Number of Threads/Ranks')
plt.ylabel('Runtime/N^2 (seconds)')
plt.title('Runtime/N^2 vs Number of Threads/Ranks for N=4096')
plt.grid(True, which="both", ls="--")
plt.legend()
plt.show()
```
